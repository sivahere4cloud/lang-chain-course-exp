## Define a model and create an object

```python
from pydantic import BaseModel

class JobPosting(BaseModel):
    title: str
    company: str
    salary: int


job = JobPosting(title="AI Engineer",company="Hyra",salary="90000")

print(job)
print(job.title)
print(job.model_dump())
print(type(job.model_dump()))
```

```python
#coercion (Pydantic fixes compatible data)
job2 = JobPosting(title="AI Engineer", company="Hyra", salary="90000")

print(job2.salary)
print(type(job2.salary))
```

```python
#validation errors (Pydantic rejects bad data)
from pydantic import ValidationError

try:
    bad_job = JobPosting(title="AI Engineer", salary="ninety thousand")
except ValidationError as error:
    print(error)
```

```python
class JobPostingV2(BaseModel):
    title: str
    company: str
    salary: int | None = None


job3 = JobPostingV2(title="Generative AI Engineer", company="PortF")

print(job3)
print(job3.salary)
```

```python
from pydantic import BaseModel, ValidationError

bad_data = {"title": "AI Engineer", "compnay": "Hyra", "salary": "ninety thousand"}

print("Plain dict accepted it:", bad_data)

class JobPosting(BaseModel):
    title: str
    company: str
    salary: int

try:
    job = JobPosting.model_validate(bad_data)
except ValidationError as error:
    print("Pydantic rejected it:")
    print(error)
```
